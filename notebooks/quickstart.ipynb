# MRAS quickstart

One noise level of the Darcy benchmark at a coarse mesh, then the same on a
finer mesh. Everything else (mesh, spaces, data feed, solver) comes from
the package. For the full paper runs use the scripts in `scripts/`.

```python
import matplotlib

from mras.problems.darcy import DarcyProblem
from mras.plotting import error_figure, evolution_figure
```

```python
prob = DarcyProblem(h=0.08, dt=1e-3, tend=0.05, seed=0)
results = prob.run(levels=[0.0, 0.05])
```

```python
for r in results:
    print(f"noise {r.noise_level:g}: err_u={r.final_error_u:.4f}  "
          f"err_q={r.final_error_q:.4f}")
```

```python
error_figure(results, prob.dt)
```

```python
evolution_figure([[r.state.gfu for r in results],
                  [r.state.gfq for r in results]], prob.box,
                 col_labels=[f"level {r.noise_level:g}" for r in results],
                 row_labels=["u", "q"])
```
